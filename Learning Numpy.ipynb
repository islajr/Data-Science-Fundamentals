MATRICES IN NUMPY.
With the **.matrix** function, we can create a matrix out of normal data-types.

At first, it gives the same output as that of **.array**, but the matrix advantage lies in the fact that
more specific matrix operations can now be applied on them.
See example below:

```python
mat = np.matrix([[1, 2, 3, 4], [5, 6, 7, 8]])
print(mat)

# Other ways of creating a matrix include:
m = np.matrix('1 3 5 7; 2 4 6 8')   # syntax: no commas, separated by semi-colons.
print(m)
```

See below some special numpy functions that work only with matrices.

```python
print(np.diagonal(m))   # printing only the elements on the diagonal.
```

Other array functions (min, max, argmin, etc) also work in matrices.

```python
print(m.max())
print(m.argmin())
```

Same also goes for operational functions between matrices; adding, subtracting and all are also possible.

```python
print(mat + m)
```

All in all, matrices are just a much better alternative to arrays if a programmer wants to work with much more complex datasets.

SPLITTING AN ARRAY

Using **hsplit()**, you can split an array along its horizontal axis, either by specifying the number of equally shaped arrays to return, or by specifying the columns after which the division should occur.

You can also split an array with the **.split()** function. The **.split()** function takes in two arguments: the array name and the number of parts to which it should split.

```python
split = np.arange(10)
print(split)

print(np.hsplit(split, 2))
print(np.split(split, 2))
print(np.split(split, [4, 7]))  # telling the array to split firstly at index #4 and split again at #7
```

INCREASING THE DIMENSION LEVEL OF ARRAYS.

**.newaxis** will increase the dimension of your array by one when used once.
    In other words, 1D arrays will become 2D, 2D becomes 3D and so on.

You can explicitly convert a 1D array with either a row vector or a column vector using **np.newaxis**.

```python
dimA = np.array([1, 2, 3, 4, 5])
print(np.shape(dimA))

dimB = dimA[np.newaxis, :]  # with this, a new axis has been added
print(np.shape(dimB))
```

You can also expand an array by inserting a new axis at a specified position with **np.expand_dims**.

```python
expand = np.expand_dims(dimA, axis=1)   # adding an axis at index #1
print(expand.shape)
```

You can also use **np.nonzero()** to select elements or indices from an array

ITERATING OVER AN ARRAY

This is done with the **.nditer()** function.

```python
a = np.arange(0, 45, 5)
a = a.reshape(3, 3)

for x in np.nditer(a, order='c'):   # the way the array is expected to be iterated can be set with this 'order' feature; either the c-style or the fortran-style.
    print(x)
```
