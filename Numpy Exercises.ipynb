To test if none of the elements given in an array is zero.

```python
a = np.array([0, 1, 2, 3, 4])
print(np.all(a))

b = np.array([1, 2, 3, 4])
print(np.all(b))

# if it returns 'False', then the array contains zero, if it returns 'True', the array doesn't contain zero.
```

To test whether any of the elements in an array is non-zero.

```python
c = np.array([1, 0, 0, 0])
print(np.any(c))

d = np.array([0, 0, 0, 0])
print(np.any(d))

# if it returns 'false', then the array doesn't contain a non-zero element but if it returns 'true', then the array contains a non-zero element.

# you can also loop through the array manually to get the same result
for i in np.nditer(c):
    if i != 0:
        print(f'{c} contains a non-zero.')
        break
    else:
        print(f'{c} does not contain a non-zero value.')
```

To create an element-wise comparison of two given arrays.

```python
one = np.array([3, 5])
two = np.array([2, 5])

# comparison - greater than:
print(np.greater(one, two))
# comparison - less
print(np.less(one, two))

# comparison - greater or equal
print(np.greater_equal(one, two))

# comparison = lesser or equal
print(np.less_equal(one, two))
```

To create a 10x10 matrix in which all the elements on the border will be equal to 1 and the rest, 0.

```python
superIdentity = np.zeros(100).reshape(10, 10)
# loop through the corners through indexing and replace the numbers with 1.
for i in range(0, 10):
    superIdentity[i, 0] = 1
    superIdentity[i, 9] = 1
    superIdentity[0, i] = 1
    superIdentity[9, i] = 1
print(superIdentity)

print() # blank space
# an easier method:
superIdentity[1:-1, 1:-1] = 0   # setting all the rows to 0 and all the columns to 0
print(superIdentity)
```

To create a 5x5 zero matrix where elements on the main diagonal equal 1, 2, 3, 4, 5

```python
zeroM = np.zeros((5, 5))
counter = 1
for i in range(0, 5):
    zeroM[i, i] = counter
    counter += 1
print(zeroM)

#an easier method.
zeroMM = np.diag([1, 2, 3, 4, 5])
print(zeroMM)
""" when fixing the diagonals, it isn't necessary to define the array or set the size
of the array as the rest of the array is automatically filled with the zeros and the
size is set to be relative to the number of diagonals provided.
"""
```

To create a 4x4 matrix in which 0 and 1 are staggered with zeros on the diagonal.

```python
stags = np.zeros((4, 4))
stags[::2, 1::2] = 1
stags[1::2, ::2] = 1
print(stags)
```
