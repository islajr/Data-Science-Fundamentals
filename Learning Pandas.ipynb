# Pandas Series Data Structure

A pandas series is a one dimensional array showing the containing values and their corresponding indexes.
It is useful for displaying the contents of lists, arrays, tuples and dictionaries.
It also shows the data type of elements passed through it.

```python
print(pd.Series(['Isla', 'Archer', 19]))
#extreme attention should be paid to the syntax, particularly the use of brackets and square brackets.
```

```python
"""
Whenever the .series method is called on a dictionary, it converts the dictionary keys into indexes and attaches them to the values. The dtype is also set to 'object'
"""

dict_check = pd.Series({
    'Name': 'Isla',
    'Age': 19,
    'Hobbies': 'Programming'

})

print(dict_check)

# We can also get the index of a panda series by using the s.index() method.

print(dict_check.index)

# Alternately, using the .values method fetches the values of the series.
print(dict_check.values)
```

```python
# we can also do this: define the values of a series and its index separately from one another.

print(pd.Series(['Isaac', 'Akinmokun', 'Oluwaseun'], index = ['First name', 'Middle name', 'Last Name']))
```

# Querying a Panda Series.
To query a panda series is to fetch its values externally either through indexing or by calling its key(index name).

We query a panda series with the **.iloc()** method for *integer loc* and with the **.loc()** method.

**.iloc()** takes in an integer that is always the index number of the intended value.
**.loc()** takes in an argument that is the index name of the intended value.

```python
query1 = pd.Series({
    'Nigeria': 'Naira',
    'USA': 'Dollar',
    'UK': 'Pound Sterling'
})

# querying the above series.
# Note: it is covered in square brackets because it is basically indexing.
print(query1.loc['Nigeria'])    #.location example.
print(query1.iloc[0])   #.integer location example.
```

# Adding to a Panda Series
We can also add to a panda series with the query method.
Simply use either of the **.loc()** or **.iloc()** methods as appropriate, fix in your key and assign it to a value and the panda series will be updated.

```python
query1.loc['Italy'] = 'Euros'
print(query1)
```

# Pandas Dataframe
While a panda series was a single dimensional array, the pandas dataframe is a multidimensional array showing the containing values and their respective indexes.

**The Dataframe Data Structure**
The dataframe is invoked when the data has become too complex for the panda series to contain.
Pandas dataframe takes in different panda series as its values in order to deal with them as one big dataset.

It can be accessed with the **.Dataframe()** method.

```python
Person1 = pd.Series({
    'Name': 'Isaac',
    'Age': 19,
    'Gender': 'Male',
    'Height': 182
})

Person2 = pd.Series({
    'Name': 'Emmanuel',
    'Age': 17,
    'Gender': 'Male',
    'Height': 176,
    'School': 'University of Ilorin'
})

Persons = pd.DataFrame([Person1, Person2], index = ['First', 'Second'])
```

As seen above, dataframes presents an organized way to work on panda series of a similar nature.

In the event that not all series share the same index, dataframe registers a NaN value in place of the missing data.

**Querying a Dataframe.**

Dataframes are queried in pretty much the same way Series are queried.
See below:

```python
# location querying:
print(Persons.loc['First'])
print()
print(f"{Persons.loc['First']['Name']}'s height is: {Persons.loc['First']['Height']}cm.")

print()

# integer location querying:
print(Persons.iloc[1])
print()
print(f"{Persons.iloc[1][0]}'s height is {Persons.iloc[1][3]}cm.")

print()

# Commas also work in case the brackets seem a little ugly.
print(Persons.loc['First', 'Name'])
```

Dataframes, just like matrices, can be transposed with the **.T** feature.
This may be particularly useful when one wants to locate elements with their more familiar key instead of the given index in order to produce results of the same type.

```python
print(Persons.T.loc['Age'])
```

```python
# This is also another way to use .loc

print(Persons.loc[:, ['Name', 'Age']])

"""
The colon part of the code above fetches data from everywhere in the dataframe.
We then proceed to extract only the values with the defined keys, i.e: Name and Age.
"""
```

**Dropping rows and columns from a dataframe**

The need may arise for a data scientist to either temporarily or permanently remove from axes of data from a dataset; this is what *dropping* is all about.

It is achieved by using the **.drop()** method.

The **.drop()** method in its base form is used to drop rows.
If one wants to drop columns, then axis must be specified

```python
print(Persons.drop(['Second'])) # dropping a row
print()
print(Persons.drop('Name', axis=1)) # dropping a column

# If by for any reason a developer wants to permanently drop a row from a dataset, all he need do is to add the *inplace=True* feature and the change will be permanent.
```

**Dataframe Indexing and Loading**

```python
# It is very possible to obtain the data in a column without using the .T or .loc() features. This can be done directly as seen below:

print(Persons['Height'])

print()

# We can then choose to be that bit more specific with the .loc() feature in order to narrow it down to a particular index.

print(Persons['Height'].loc['First'])
```
